### Regex

| Regex | Significado | Exemplo |
| :--- | :--- | :--- |
| `\d` | Número | `5` |
| `\D` | Tudo que não é número | `A`, `-`, espaço |
| `\w` | Letra, número ou `_` | `Python`, `10` |
| `\s` | Espaço em branco | espaço |
| `.` | Praticamente qualquer caractere | `A`, `5`, `@` |
| `+` | Uma ou mais ocorrências | `123` |
| `*` | Zero ou mais ocorrências | — |
| `?` | Zero ou uma ocorrência | — |
| `[0-9]` | Números de 0 a 9 | `7` |
| `[A-Za-z]` | Letras | `Python` |
| `^` | Início do texto | — |
| `$` | Final do texto | — |

```python
import re
```

```python
texto = "Pedido 4587"

resultado = re.findall(r"\d", texto)

print(resultado)

print(type(resultado))
```

```python
for i in resultado:
    print (i * 2)
```

```python
texto = "Pedido 4587"

# Lista
resultado = re.findall(r"\d+", texto)

print(resultado)

print(type(resultado))
```

```python
# Tudo que não é número
texto = "ABCD123"

resultado = re.findall(r"\D", texto)

print(resultado)

print(type(resultado))
```

```python
# Tudo que não é número
texto = "ABCD123"

# Lista 
resultado = re.findall(r"\D+", texto)

print(resultado)

print(type(resultado))
```

```python
telefone = "(55) 99999-9999"

# função de substituição
telefone_limpo = re.sub(r"\D", "", telefone)

print(telefone_limpo)
```

```python
# \w => letras, números e "_"
texto = "Python_2026!"

# Retirou o ponto de exclamação (!)
resultado = re.findall(r"\w", texto)

print(resultado)
```

```python
# \w => letras, números e "_"
texto = "Python_2026!"

# Retirou o ponto de exclamação (!)
resultado = re.findall(r"\w+", texto)

print(resultado)
```

```python
# \s => espaço em branco
texto = " Ana Silva "

resultado = re.findall(r"\s", texto)

print(resultado)
```

```python
nome = "Ana             Silva"

# função de substituição
nome_limpo = re.sub(r"\s+", " ", nome)

# string
print(nome_limpo)

print(type(nome_limpo))
```

```python
nome = "Ana     R.        Silva"

# função de substituição
nome_limpo = re.sub(r"\s+", " ", nome)

# string
print(nome_limpo)

print(type(nome_limpo))
```

```python
# . => praticamente qualquer caractere
texto = "gato peixe sapo"

# lê qualquer caractere que vem antes + "to" 
resultado = re.findall(r".to", texto)

print(resultado)

# Lista 
print(type(resultado))
```

```python
# . => praticamente qualquer caractere
texto = "gato peixe sapato"

# lê apenas um caractere que vem antes + "to" 
resultado = re.findall(r".ato", texto)

print(resultado)

# Lista 
print(type(resultado))
```

```python
texto = "gato peixe sapo"

# lê gat + um caractere que vem depois
resultado = re.findall(r"gat.", texto)

print(resultado)
```

```python
cep = "90000-500"
# cep = "90000--500"

# ? => zero ou uma ocorrência, caso contrário resulta em erro
resultado = re.search(r"\d{5}-?\d{3}", cep)
print(resultado.group())

cep = "90000500"
resultado2 = re.search(r"\d{5}-?\d{3}", cep)
print(resultado2.group())
```

#### Quantificadores Regex

- `*` -> o elemento pode aparecer **zero ou mais vezes**
- `?` -> o elemento pode aparecer **zero ou uma vez**
- `+` -> o elemento deve aparecer **uma ou mais vezes**

#### Exemplos

- `\s*` -> nenhum espaço ou vários espaços
- `-?` -> hífen opcional
- `\d+` -> um ou mais números

```python
# ^ -> início do texto
# O padrão precisa estar no início do texto

codigo = "PROD123"

resultado = re.search(r"^PROD", codigo)

print(resultado)
print(resultado.group())

# Classe re.Match 
print(type(resultado))

# string
print(type(resultado.group()))
```

```python
# O padrão ^ precisa estar no início do texto

codigo = "123PROD"

resultado = re.search(r"^PROD", codigo)

print(resultado)

# Classe NoneType
print(type(resultado))

```

```python
# $ -> final do texto
# Queremos verificar se um arquivo termina em .csv

arquivo = "vendas.csv"

resultado = re.search(r"\.csv$", arquivo)

print(resultado)

print(resultado.group())
```
